```python
import cv2
import numpy as np
from collections import defaultdict
from ultralytics import YOLO

best_model = YOLO('runs/detect/train/weights/best.pt')
video_path = 'test video.mp4'
output_path = 'tracked_result_final.mp4'

# 1. 영상 중간 프레임 5장을 뽑아 어느 회전 방향에서 물체가 가장 잘 잡히는지 자동 탐색
cap = cv2.VideoCapture(video_path)
total = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
sample_indices = np.linspace(0, max(0, total - 1), 5, dtype=int)
sample_frames = []
for idx in sample_indices:
    cap.set(cv2.CAP_PROP_POS_FRAMES, idx)
    ret, f = cap.read()
    if ret:
        sample_frames.append(f)
cap.release()

rotations = {
    "원본(회전 없음)": None,
    "시계방향 90도 회전": cv2.ROTATE_90_CLOCKWISE,
    "반시계방향 90도 회전": cv2.ROTATE_90_COUNTERCLOCKWISE,
    "180도 회전": cv2.ROTATE_180
}

best_rot_name = "원본(회전 없음)"
best_rot_code = None
max_hits = -1

for name, rot_code in rotations.items():
    hits = 0
    for f in sample_frames:
        img = cv2.rotate(f, rot_code) if rot_code is not None else f
        res = best_model.predict(img, conf=0.05, imgsz=512, verbose=False)
        if res[0].boxes is not None:
            hits += len(res[0].boxes)
    print(f"[{name}] 샘플 탐지 개수: {hits}개")
    if hits > max_hits:
        max_hits = hits
        best_rot_name = name
        best_rot_code = rot_code

print(f"\n==> 선택된 최적 방향: {best_rot_name} (총 {max_hits}개 탐지)")

# 2. 선택된 최적 방향으로 전체 영상 추적 및 궤적 그리기
cap = cv2.VideoCapture(video_path)
fps = int(cap.get(cv2.CAP_PROP_FPS)) or 30
out = None
track_history = defaultdict(lambda: [])
detected_frames = 0
total_frames = 0

while cap.isOpened():
    success, frame = cap.read()
    if not success:
        break
    total_frames += 1

    if best_rot_code is not None:
        frame = cv2.rotate(frame, best_rot_code)

    if out is None:
        h, w = frame.shape[:2]
        out = cv2.VideoWriter(output_path, cv2.VideoWriter_fourcc(*'mp4v'), fps, (w, h))

    results = best_model.track(frame, persist=True, conf=0.05, iou=0.5, imgsz=512, verbose=False)

    if results[0].boxes is not None and len(results[0].boxes) > 0:
        detected_frames += 1
        annotated_frame = results[0].plot()

        if results[0].boxes.id is not None:
            boxes = results[0].boxes.xywh.cpu()
            track_ids = results[0].boxes.id.int().cpu().tolist()

            for box, track_id in zip(boxes, track_ids):
                x, y, _, _ = box
                track = track_history[track_id]
                track.append((float(x), float(y)))
                if len(track) > 60:
                    track.pop(0)

                points = np.hstack(track).astype(np.int32).reshape((-1, 1, 2))
                cv2.polylines(annotated_frame, [points], isClosed=False, color=(0, 255, 255), thickness=4)

        out.write(annotated_frame)
    else:
        out.write(frame)

cap.release()
if out is not None:
    out.release()
print(f"최종 완료! 총 {total_frames}프레임 중 {detected_frames}프레임에서 객체 탐지 성공!")
```
